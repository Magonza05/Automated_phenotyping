#MAE and RMSE

```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Group the data by plant
grouped_data = data.groupby('Plant Name')

# Initialize lists to store MAE and RMSE for each plant
mae_per_plant = []
rmse_per_plant = []

# Iterate over each plant
for plant, plant_data in grouped_data:
    # Extract the original counts and generated counts
    original_counts = plant_data['Original Count'].tolist()
    generated_counts = plant_data['Generated Count'].tolist()

    # Calculate MAE for the current plant
    mae = mean_absolute_error(original_counts, generated_counts)
    mae_per_plant.append(mae)

    # Calculate RMSE for the current plant
    rmse = np.sqrt(mean_squared_error(original_counts, generated_counts))
    rmse_per_plant.append(rmse)

    # Plot the line plot for the current plant
    plt.plot(plant_data['Date_Month_Year'], plant_data['Generated Count'], label=plant)

# Set the plot labels and title
plt.xlabel('Date')
plt.ylabel('Generated Count')
plt.title('Generated Count for Each Plant')
plt.legend()
plt.xticks(rotation=45)

# Show the plot
plt.show()

# Print the MAE and RMSE for each plant
for i, plant in enumerate(grouped_data.groups.keys()):
    print(f"Plant {plant}: MAE={mae_per_plant[i]}, RMSE={rmse_per_plant[i]}")
```

##MAE and RMSE -- 2nd Approach

```python
# Calculate RMSE and MAE for each plant
rmse_values = []
mae_values = []

for plant in plants:
    plant_data = data1[data1['Plant '] == plant]
    original_counts = plant_data['Original Count']
    generated_counts = plant_data['Generatted Count']

    rmse = np.sqrt(np.mean((original_counts - generated_counts) ** 2))
    mae = np.mean(np.abs(original_counts - generated_counts))

    rmse_values.append(rmse)
    mae_values.append(mae)

# Plot RMSE and MAE for each plant
plt.figure(figsize=(10, 6))
plt.plot(plants, rmse_values, label='RMSE')
plt.plot(plants, mae_values, label='MAE')
plt.xlabel('Plant Name')
plt.ylabel('Error')
plt.title('RMSE and MAE for Each Plant')
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
for plant, rmse, mae in zip(plants, rmse_values, mae_values):
    print(f"Plant {plant}: RMSE = {rmse}, MAE = {mae}")
```
