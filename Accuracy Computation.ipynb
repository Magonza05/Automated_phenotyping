```python
import pandas as pd
import numpy as np
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
# Load the original count and generated count files into separate DataFrames
df_original = pd.read_excel('/content/drive/MyDrive/Results_lettuce csv/Leaf_count_original.xlsx')
df_generated = pd.read_excel('/content/drive/MyDrive/Results_lettuce csv/Leaf_count_Method1_sorted.xlsx')
```

```python
 df_original.shape, df_generated.shape
```

```python
# Merge the original and generated DataFrames based on common columns
merged_df = df_original.merge(df_generated, on=['Plant Name', 'Date_Month_Year', 'Time'])

# Compute the MAE and RMSE for each plant
plants = ['Plant 1', 'Plant 2', 'Plant 3', 'Plant 4', 'Plant 5', 'Plant 6']
for plant in plants:
    # Filter the merged DataFrame for the current plant
    filtered_df = merged_df[merged_df['Plant Name'] == plant]

    # Extract the original count and generated count for the current plant
    original_count = filtered_df['Leaf_Count_x'].values
    generated_count = filtered_df['Leaf_Count_y'].values

    # Compute the MAE and RMSE for the current plant
    mae = np.mean(np.abs(original_count - generated_count))
    rmse = np.sqrt(np.mean((original_count - generated_count) ** 2))

    # Print the results
    print(f"Plant: {plant}")
    print(f"MAE: {mae}")
    print(f"RMSE: {rmse}")
    print()
```

```python
# Initialize lists to store the MAE and RMSE values for each plant
mae_values = []
rmse_values = []

# Compute the MAE and RMSE for each plant and store the values in the lists
for plant in plants:
    filtered_df = merged_df[merged_df['Plant Name'] == plant]
    original_count = filtered_df['Leaf_Count_x'].values
    generated_count = filtered_df['Leaf_Count_y'].values
    mae = np.mean(np.abs(original_count - generated_count))
    rmse = np.sqrt(np.mean((original_count - generated_count) ** 2))
    mae_values.append(mae)
    rmse_values.append(rmse)

# Create a bar chart for MAE
plt.figure(figsize=(10, 5))
plt.bar(plants, mae_values)
plt.xlabel('Plant')
plt.ylabel('MAE')
plt.title('Mean Absolute Error (MAE) for Each Plant')
plt.show()

# Create a bar chart for RMSE
plt.figure(figsize=(10, 5))
plt.bar(plants, rmse_values)
plt.xlabel('Plant')
plt.ylabel('RMSE')
plt.title('Root Mean Squared Error (RMSE) for Each Plant')
plt.show()
```

##Plot the graph and visualize

```python
# Select the plants
plants = ['Plant 1', 'Plant 2', 'Plant 3', 'Plant 4', 'Plant 5', 'Plant 6']

# Calculate the number of rows and columns for the grid layout
num_rows = 2
num_cols = 3

# Create a grid of subplots for each plant
fig, axes = plt.subplots(num_rows, num_cols, figsize=(12, 8))
fig.tight_layout(pad=3.0)

# Define the width of the bars
bar_width = 0.4

# Iterate over the plants and plot in the grid
for i, plant in enumerate(plants):
    # Compute the row and column index for the current plant
    row_idx = i // num_cols
    col_idx = i % num_cols

    # Filter the merged DataFrame for the current plant
    filtered_df = merged_df[merged_df['Plant Name'] == plant]

    # Extract the leaf count for the original and generated count
    original_count = filtered_df['Leaf_Count_x']
    generated_count = filtered_df['Leaf_Count_y']

    # Compute the mean leaf count for original and generated count
    mean_original_count = np.mean(original_count)
    mean_generated_count = np.mean(generated_count)

    # Calculate the percentage of correctness
    percentage_correct = (mean_generated_count / mean_original_count) * 100

    # Plot the bar chart in the current subplot
    ax = axes[row_idx, col_idx]
    ax.bar([0], [mean_original_count], width=bar_width, label='Original')
    ax.bar([bar_width], [mean_generated_count], width=bar_width, label='Generated')
    ax.set_xticks([0, bar_width])
    ax.set_xticklabels(['Original', 'Generated'])
    ax.set_xlabel('Count Type')
    ax.set_ylabel('Mean Leaf Count')
    ax.set_title(f'Leaf Count Comparison - {plant}')
    ax.text(0, mean_original_count, f'{mean_original_count:.2f}', ha='center', va='bottom')
    ax.text(bar_width, mean_generated_count, f'{mean_generated_count:.2f} ({percentage_correct:.2f}%)', ha='center', va='bottom')
    ax.set_ylim(0, max(mean_original_count, mean_generated_count) * 1.1)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.legend()

# Hide any empty subplots
if len(plants) < num_rows * num_cols:
    for i in range(len(plants), num_rows * num_cols):
        row_idx = i // num_cols
        col_idx = i % num_cols
        axes[row_idx, col_idx].axis('off')

# Adjust the spacing between subplots
plt.subplots_adjust(hspace=0.5)

# Show the plot
plt.show()
```

```python
# Select the date for plotting
selected_date = '30_05_2023'

# Filter the data for the selected date
filtered_original = df_original[df_original['Date_Month_Year'] == selected_date]
filtered_generated = df_generated[df_generated['Date_Month_Year'] == selected_date]

# Extract the unique recorded times
recorded_times = np.sort(filtered_original['Time'].unique())

# Create lists to store the mean leaf count for original and generated count
mean_original_counts = []
mean_generated_counts = []

# Compute the mean leaf count for each recorded time
for time in recorded_times:
    # Filter the data for the current time
    filtered_original_time = filtered_original[filtered_original['Time'] == time]
    filtered_generated_time = filtered_generated[filtered_generated['Time'] == time]

    # Extract the leaf count for the original and generated count
    original_count = filtered_original_time['Leaf_Count']
    generated_count = filtered_generated_time['Leaf_Count']

    # Compute the mean leaf count for original and generated count
    mean_original_count = np.mean(original_count)
    mean_generated_count = np.mean(generated_count)

    # Append the mean leaf counts to the respective lists
    mean_original_counts.append(mean_original_count)
    mean_generated_counts.append(mean_generated_count)

# Plot the bar chart
plt.figure(figsize=(10, 6))
bar_width = 0.35
index = np.arange(len(recorded_times))

# Plot the original count bars
plt.bar(index, mean_original_counts, bar_width, label='Original Count')

# Plot the generated count bars
plt.bar(index + bar_width, mean_generated_counts, bar_width, label='Generated Count')

# Configure the x-axis
plt.xticks(index + bar_width / 2, recorded_times)
plt.xlabel('Time')
plt.ylabel('Leaf Count')
plt.title(f'Leaf Count Comparison - Date: {selected_date}')
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```
