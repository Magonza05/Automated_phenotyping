#Trying the resizing and quadrant dividation via loop

```python
import os
from PIL import Image

# Define the folder paths
input_folder = "/content/drive/MyDrive/Balcony Images"
output_folder = "/content/drive/MyDrive/Balcony Cropped and Quadrants"

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Iterate through the images in the input folder
for image_file in os.listdir(input_folder):
    # Extract the date from the image filename
    date_str = image_file.split('_')[1] + '_' + image_file.split('_')[2]  # Assuming the date is in the format 'day_month_year'

    # Check if the date falls within the specified range
    if "26_06_2023" <= date_str <= "29_06_2023":
        continue  # Skip this image if the date is within the range

    # Open the image
    image_path = os.path.join(input_folder, image_file)
    image = Image.open(image_path)

    # Crop the image
    cropped_image = image.crop((620, 110, 620+1350, 110+1090))

    # Divide the image into four quadrants
    width, height = cropped_image.size
    quadrant_width = width // 2
    quadrant_height = height // 2

    quadrants = []
    for j in range(2):
        for i in range(2):
            left = i * quadrant_width
            upper = j * quadrant_height
            right = left + quadrant_width
            lower = upper + quadrant_height
            quadrant = cropped_image.crop((left, upper, right, lower))
            quadrants.append(quadrant)

    # Generate the output filenames for cropped image and quadrants
    plant_name = image_file.split('_')[0]  # Extract the plant name from the original image file name
    quadrant_filenames = [
        f"plant1_{image_file}",
        f"plant2_{image_file}",
        f"plant3_{image_file}",
        f"plant4_{image_file}"
    ]

    # Save the cropped image and quadrants
    for i, quadrant in enumerate(quadrants):
        quadrant_path = os.path.join(output_folder, quadrant_filenames[i])
        quadrant.save(quadrant_path)

print("Image cropping and dividing into quadrants completed.")
```

##Trying babyroom Rotation, resizing and cropping the images

```python
import os
import cv2
from PIL import Image

# Define the folder paths
input_folder = "/content/drive/MyDrive/Babyroom Images "
output_folder = "/content/drive/MyDrive/Babyroom Cropped and Quadrants"

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Iterate through the images in the input folder
for image_file in os.listdir(input_folder):
    # Open the image
    image_path = os.path.join(input_folder, image_file)
    image = cv2.imread(image_path)

    # Get the dimensions of the image
    height, width = image.shape[:2]

    # Calculate the rotation angle in degrees (rotate by 4 degrees)
    angle = 4

    # Calculate the rotation matrix
    rotation_matrix = cv2.getRotationMatrix2D((width/2, height/2), angle, 1)

    # Apply the rotation to the image
    rotated_image = cv2.warpAffine(image, rotation_matrix, (width, height))

    # Define the cropping coordinates
    x = 350
    y = 160
    w = 1400
    h = 1050

    # Crop the rotated image
    cropped_image = rotated_image[y:y+h, x:x+w]

    # Convert the image to PIL format for easier saving
    cropped_pil_image = Image.fromarray(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))

    # Divide the cropped image into four quadrants
    quadrant_width = w // 2
    quadrant_height = h // 2

    quadrants = []
    for j in range(2):
        for i in range(2):
            left = i * quadrant_width
            upper = j * quadrant_height
            right = left + quadrant_width
            lower = upper + quadrant_height
            quadrant = cropped_pil_image.crop((left, upper, right, lower))
            quadrants.append(quadrant)

    # Generate the output filenames for quadrants
    plant_names = ["plant5", "plant6", "plant7", "plant8"]
    quadrant_filenames = [
        f"{plant_names[0]}_{image_file}",
        f"{plant_names[1]}_{image_file}",
        f"{plant_names[2]}_{image_file}",
        f"{plant_names[3]}_{image_file}"
    ]

    # Save the quadrants
    for i, quadrant in enumerate(quadrants):
        quadrant_path = os.path.join(output_folder, quadrant_filenames[i])
        quadrant.save(quadrant_path)

print("Image rotation, cropping, and dividing into quadrants completed.")
```

```python
#Defining the height, width of the image

x= 150
y= 120

w=1600
h=1300


cropped_image = rgb_image[y:y+h, x:x+w] # Cropping using Slicing
#plt.axis("off")
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.show() # Display the original image
print(cropped_image.shape)
```

```python
# Define the four corners of the desired output ROI
output_size = (w, h)  # Same size as the cropped region
output_corners = np.array([(0, 0), (output_size[0], 0), (0, output_size[1]), output_size], dtype=np.float32)

# Define the corners of the cropped image
cropped_corners = np.array([(0, 0), (w, 0), (0, h), (w, h)], dtype=np.float32)

# Calculate the perspective transformation matrix
matrix = cv2.getPerspectiveTransform(cropped_corners, output_corners)

# Apply the perspective transformation to the cropped image
transformed_image = cv2.warpPerspective(cropped_image, matrix, output_size)

# Divide the transformed image into four quadrants
height, width = transformed_image.shape[:2]
quadrant_width = width // 2
quadrant_height = height // 2

quadrant1 = transformed_image[:quadrant_height, :quadrant_width]
quadrant2 = transformed_image[:quadrant_height, quadrant_width:]
quadrant3 = transformed_image[quadrant_height:, :quadrant_width]
quadrant4 = transformed_image[quadrant_height:, quadrant_width:]

# Display the original cropped image and the quadrants
plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))
plt.title("Cropped Image")

plt.subplot(2, 3, 2)
plt.imshow(cv2.cvtColor(quadrant1, cv2.COLOR_BGR2RGB))
plt.title("Quadrant 1")

plt.subplot(2, 3, 3)
plt.imshow(cv2.cvtColor(quadrant2, cv2.COLOR_BGR2RGB))
plt.title("Quadrant 2")

plt.subplot(2, 3, 4)
plt.imshow(cv2.cvtColor(quadrant3, cv2.COLOR_BGR2RGB))
plt.title("Quadrant 3")

plt.subplot(2, 3, 5)
plt.imshow(cv2.cvtColor(quadrant4, cv2.COLOR_BGR2RGB))
plt.title("Quadrant 4")

plt.tight_layout()
plt.show()
```

```python

folder_path = '/content/drive/MyDrive/Lettuce image'  # Specify the path to the "cropped" folder
output_folder_path = '/content/drive/MyDrive/Lettuce image/Cropped '  # Specify the path to the output folder

# Get a list of all image files in the folder
image_files = [file for file in os.listdir(folder_path) if file.startswith('cropped_')]

# Loop through each image file
for image_file in image_files:
    # Extract the date from the image filename
    date_str = image_file.split('_')[1]

    # Check if the date falls within the specified range
    if '26_05_2023_00' <= date_str <= '29_05_2023_13':
        # Construct the path to the image file
        image_path = os.path.join(folder_path, image_file)

        # Load the image
        image = io.imread(image_path)

        # Rotate the image to the left (counter-clockwise)
        rotated_image = transform.rotate(image, 90, resize=True)

        # Generate the output filename
        output_filename = image_file.replace('cropped_', 'rotated_')

        # Construct the path to the output file
        output_path = os.path.join(output_folder_path, output_filename)

        # Save the rotated image
        io.imsave(output_path, rotated_image)
```

#Cropping the actual image in four quadrants

```python
# Specify the folder path
folder_path = '/content/drive/MyDrive/Lettuce image'
output_folder = '/content/drive/MyDrive/Final_lettuce_balcony'

# Create the output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)

# Get the list of files in the folder
files = os.listdir(folder_path)

# Iterate over the files
for file_name in files:
    # Skip directories
    if os.path.isdir(os.path.join(folder_path, file_name)):
        continue

    # Open the image
    image_path = os.path.join(folder_path, file_name)
    image = Image.open(image_path)

    # Get the original filename
    original_filename = file_name.split('_')[1:]  # Assuming the original filename starts from the second position

    # Crop the image into four quadrants
    width, height = image.size
    half_width = width // 2
    half_height = height // 2

    quadrant1 = image.crop((0, 0, half_width, half_height))
    quadrant2 = image.crop((half_width, 0, width, half_height))
    quadrant3 = image.crop((0, half_height, half_width, height))
    quadrant4 = image.crop((half_width, half_height, width, height))

    # Save each quadrant in the output folder with the specified names
    output_folder_path = os.path.join(output_folder, original_filename[0])
    os.makedirs(output_folder_path, exist_ok=True)

    quadrant1_name = f"plant1{'_'.join(original_filename)}"
    quadrant2_name = f"plant2{'_'.join(original_filename)}"
    quadrant3_name = f"plant3{'_'.join(original_filename)}"
    quadrant4_name = f"plant4{'_'.join(original_filename)}"

    quadrant1.save(os.path.join(output_folder_path, quadrant1_name))
    quadrant2.save(os.path.join(output_folder_path, quadrant2_name))
    quadrant3.save(os.path.join(output_folder_path, quadrant3_name))
    quadrant4.save(os.path.join(output_folder_path, quadrant4_name))
```
